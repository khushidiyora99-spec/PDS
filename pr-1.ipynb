# Variables, Data Types and Operators

This program demonstrates variables, data types (int, float, str, bool) and all major Python operators.

```python
# Variables and Data Types

a = 10
b = 5.5
name = "Khushi"
flag = True

print("Integer:", a)
print("Float:", b)
print("String:", name)
print("Boolean:", flag)
```

```python
# Arithmetic Operators

print("\n--- Arithmetic Operators ---")
print("Addition:", a + 5)
print("Subtraction:", a - 5)
print("Multiplication:", a * 5)
print("Division:", a / 5)
print("Floor Division:", a // 3)
print("Modulus:", a % 3)
print("Exponent:", a ** 2)
```

```python
# Assignment Operators

print("\n--- Assignment Operators ---")
        x = 10

x += 5
print("x += 5:", x)

x -= 2
print("x -= 2:", x)

x *= 2
print("x *= 2:", x)

x /= 2
print("x /= 2:", x)

x //= 2
print("x //= 2:", x)

x %= 3
print("x %= 3:", x)
```

```python
# Comparison Operators

print("\n--- Comparison Operators ---")
        p = 10
q = 20

print("p == q:", p == q)
print("p != q:", p != q)
print("p > q:", p > q)
print("p < q:", p < q)
print("p >= q:", p >= q)
print("p <= q:", p <= q)
```

```python
# Logical Operators

print("\n--- Logical Operators ---")
print("AND:", p < q and q > 15)
print("OR:", p > q or q > 15)
print("NOT:", not(p > q))
```

```python
# Membership Operators

print("\n--- Membership Operators ---")

numbers = [10, 20, 30, 40]

print("20 in numbers:", 20 in numbers)
print("50 in numbers:", 50 in numbers)
print("50 not in numbers:", 50 not in numbers)
```

```python
# Identity Operators

print("\n--- Identity Operators ---")

x = [1, 2, 3]
y = x
z = [1, 2, 3]

print("x is y:", x is y)
print("x is z:", x is z)
print("x is not z:", x is not z)
```

```python
# Bitwise Operators

print("\n--- Bitwise Operators ---")
        m = 5
n = 3

print("m & n:", m & n)
print("m | n:", m | n)
print("m ^ n:", m ^ n)
print("~m:", ~m)
print("m << 1:", m << 1)
print("m >> 1:", m >> 1)
```
